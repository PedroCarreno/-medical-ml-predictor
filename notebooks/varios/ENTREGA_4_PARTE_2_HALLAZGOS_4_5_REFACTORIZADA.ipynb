```python
# ========================================
# CONFIGURACIÓN INICIAL
# ========================================
# Solo ejecuta esto si NO has ejecutado la Parte 1

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.ensemble import RandomForestClassifier
from xgboost import XGBClassifier
from sklearn.metrics import (
    accuracy_score, precision_score, recall_score, f1_score,
    roc_auc_score, roc_curve, confusion_matrix, classification_report
)
from IPython.display import display, HTML
import warnings
warnings.filterwarnings('ignore')

# Configuración de visualización
plt.style.use('seaborn-v0_8-darkgrid')
sns.set_palette("husl")
plt.rcParams['figure.figsize'] = (14, 8)

# Cargar datasets
df_balanced = pd.read_csv('dataset_balanced_smote_tomek.csv')
df_clean = pd.read_csv('dataset_clean_final.csv')

# Preparar datos
X = df_balanced.drop('hospital_death', axis=1)
y = df_balanced['hospital_death']

categorical_cols = X.select_dtypes(include=['object']).columns
label_encoders = {}
X_encoded = X.copy()
for col in categorical_cols:
    le = LabelEncoder()
    X_encoded[col] = le.fit_transform(X[col])
    label_encoders[col] = le

X_train, X_test, y_train, y_test = train_test_split(
    X_encoded, y, test_size=0.2, random_state=42, stratify=y
)

scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

# Entrenar modelos (silenciosamente)
rf_model = RandomForestClassifier(n_estimators=100, max_depth=15, random_state=42, n_jobs=-1)
rf_model.fit(X_train_scaled, y_train)
rf_predictions = rf_model.predict(X_test_scaled)
rf_proba = rf_model.predict_proba(X_test_scaled)[:, 1]

xgb_model = XGBClassifier(n_estimators=100, max_depth=6, learning_rate=0.1, 
                          random_state=42, eval_metric='logloss', n_jobs=-1)
xgb_model.fit(X_train_scaled, y_train)
xgb_predictions = xgb_model.predict(X_test_scaled)
xgb_proba = xgb_model.predict_proba(X_test_scaled)[:, 1]

display(HTML("""
<div style='background: linear-gradient(135deg, #11998e 0%, #38ef7d 100%); padding: 30px; border-radius: 12px; margin: 20px 0; text-align: center; box-shadow: 0 5px 15px rgba(0,0,0,0.2);'>
    <h3 style='color: white; font-size: 32px; margin: 0; font-weight: bold; text-shadow: 2px 2px 4px rgba(0,0,0,0.3);'>
        ✅ CONFIGURACIÓN COMPLETADA
    </h3>
</div>
"""))
```

```python
# ANÁLISIS: GLASGOW VS MORTALIDAD
# Calcular Glasgow total para dataset limpio
df_clean['glasgow_total'] = (
    df_clean['gcs_eyes_apache'] + 
    df_clean['gcs_motor_apache'] + 
    df_clean['gcs_verbal_apache']
)

# Análisis por rangos de Glasgow
def classify_glasgow(score):
    if score <= 8:
        return 'Crítico (≤8)'
    elif score <= 12:
        return 'Moderado (9-12)'
    else:
        return 'Leve (13-15)'

df_clean['glasgow_category'] = df_clean['glasgow_total'].apply(classify_glasgow)

glasgow_analysis = df_clean.groupby('glasgow_category').agg({
    'hospital_death': ['sum', 'count', 'mean']
}).round(4)
glasgow_analysis.columns = ['deaths', 'total', 'mortality_rate']
glasgow_analysis['mortality_pct'] = glasgow_analysis['mortality_rate'] * 100
glasgow_analysis = glasgow_analysis.reset_index()

# Extraer métricas clave
critical_mortality = glasgow_analysis[glasgow_analysis['glasgow_category'] == 'Crítico (≤8)']['mortality_pct'].values[0]
moderate_mortality = glasgow_analysis[glasgow_analysis['glasgow_category'] == 'Moderado (9-12)']['mortality_pct'].values[0]
mild_mortality = glasgow_analysis[glasgow_analysis['glasgow_category'] == 'Leve (13-15)']['mortality_pct'].values[0]

critical_total = int(glasgow_analysis[glasgow_analysis['glasgow_category'] == 'Crítico (≤8)']['total'].values[0])
critical_deaths = int(glasgow_analysis[glasgow_analysis['glasgow_category'] == 'Crítico (≤8)']['deaths'].values[0])

# Mostrar resultados
display(HTML(f"""
<div style='background: white; padding: 50px; border-radius: 15px; margin: 30px 0; box-shadow: 0 10px 25px rgba(0,0,0,0.2);'>
    <h2 style='color: #000; font-size: 42px; margin: 0 0 40px 0; font-weight: bold; text-align: center; border-bottom: 5px solid #e74c3c; padding-bottom: 20px;'>
        📊 MORTALIDAD POR NIVEL DE GLASGOW
    </h2>
    
    <div style='display: grid; grid-template-columns: repeat(3, 1fr); gap: 30px; margin: 40px 0;'>
        <!-- Glasgow Crítico -->
        <div style='background: linear-gradient(135deg, #e74c3c 0%, #c0392b 100%); padding: 40px; border-radius: 15px; box-shadow: 0 8px 20px rgba(231,76,60,0.4); text-align: center;'>
            <div style='color: white; font-size: 24px; font-weight: bold; margin-bottom: 15px;'>⚠️ CRÍTICO</div>
            <div style='color: white; font-size: 28px; font-weight: bold; margin-bottom: 20px;'>Glasgow ≤8</div>
            <div style='background: rgba(255,255,255,0.95); padding: 30px; border-radius: 12px;'>
                <div style='font-size: 72px; color: #e74c3c; font-weight: bold; text-shadow: 2px 2px 4px rgba(0,0,0,0.1);'>{critical_mortality:.1f}%</div>
                <div style='font-size: 22px; color: #000; margin-top: 15px; font-weight: bold;'>MORTALIDAD</div>
                <div style='font-size: 18px; color: #666; margin-top: 10px;'>{critical_deaths:,} / {critical_total:,} pacientes</div>
            </div>
        </div>
        
        <!-- Glasgow Moderado -->
        <div style='background: linear-gradient(135deg, #f39c12 0%, #e67e22 100%); padding: 40px; border-radius: 15px; box-shadow: 0 8px 20px rgba(243,156,18,0.4); text-align: center;'>
            <div style='color: white; font-size: 24px; font-weight: bold; margin-bottom: 15px;'>⚡ MODERADO</div>
            <div style='color: white; font-size: 28px; font-weight: bold; margin-bottom: 20px;'>Glasgow 9-12</div>
            <div style='background: rgba(255,255,255,0.95); padding: 30px; border-radius: 12px;'>
                <div style='font-size: 72px; color: #f39c12; font-weight: bold; text-shadow: 2px 2px 4px rgba(0,0,0,0.1);'>{moderate_mortality:.1f}%</div>
                <div style='font-size: 22px; color: #000; margin-top: 15px; font-weight: bold;'>MORTALIDAD</div>
                <div style='font-size: 18px; color: #666; margin-top: 10px;'>{int(glasgow_analysis[glasgow_analysis['glasgow_category'] == 'Moderado (9-12)']['deaths'].values[0]):,} / {int(glasgow_analysis[glasgow_analysis['glasgow_category'] == 'Moderado (9-12)']['total'].values[0]):,} pacientes</div>
            </div>
        </div>
        
        <!-- Glasgow Leve -->
        <div style='background: linear-gradient(135deg, #27ae60 0%, #229954 100%); padding: 40px; border-radius: 15px; box-shadow: 0 8px 20px rgba(39,174,96,0.4); text-align: center;'>
            <div style='color: white; font-size: 24px; font-weight: bold; margin-bottom: 15px;'>✅ LEVE</div>
            <div style='color: white; font-size: 28px; font-weight: bold; margin-bottom: 20px;'>Glasgow 13-15</div>
            <div style='background: rgba(255,255,255,0.95); padding: 30px; border-radius: 12px;'>
                <div style='font-size: 72px; color: #27ae60; font-weight: bold; text-shadow: 2px 2px 4px rgba(0,0,0,0.1);'>{mild_mortality:.1f}%</div>
                <div style='font-size: 22px; color: #000; margin-top: 15px; font-weight: bold;'>MORTALIDAD</div>
                <div style='font-size: 18px; color: #666; margin-top: 10px;'>{int(glasgow_analysis[glasgow_analysis['glasgow_category'] == 'Leve (13-15)']['deaths'].values[0]):,} / {int(glasgow_analysis[glasgow_analysis['glasgow_category'] == 'Leve (13-15)']['total'].values[0]):,} pacientes</div>
            </div>
        </div>
    </div>
    
    <!-- Hallazgo Crítico -->
    <div style='background: linear-gradient(135deg, #667eea 0%, #764ba2 100%); padding: 40px; border-radius: 15px; margin: 40px 0; box-shadow: 0 8px 20px rgba(0,0,0,0.2);'>
        <h3 style='color: white; font-size: 36px; margin: 0 0 30px 0; font-weight: bold; text-align: center; text-shadow: 2px 2px 4px rgba(0,0,0,0.3);'>
            🎯 HALLAZGO CRÍTICO #4
        </h3>
        <div style='background: rgba(255,255,255,0.95); padding: 35px; border-radius: 12px;'>
            <div style='text-align: center; margin-bottom: 25px;'>
                <div style='font-size: 22px; color: #000; font-weight: bold; margin-bottom: 15px;'>Factor de Riesgo Glasgow Crítico vs Leve</div>
                <div style='font-size: 88px; color: #e74c3c; font-weight: bold; text-shadow: 3px 3px 6px rgba(0,0,0,0.15);'>{critical_mortality/mild_mortality:.1f}x</div>
                <div style='font-size: 20px; color: #666; margin-top: 10px;'>Un paciente con Glasgow ≤8 tiene <strong>{critical_mortality/mild_mortality:.1f} veces más riesgo</strong> de morir</div>
            </div>
            <div style='background: #f8f9fa; padding: 25px; border-radius: 10px; margin-top: 25px;'>
                <p style='font-size: 20px; line-height: 1.9; color: #000; margin: 0; text-align: center;'>
                    <strong>🚨 Implicación clínica:</strong> La escala de Glasgow es un <strong>predictor extremadamente confiable</strong>. 
                    Pacientes con Glasgow ≤8 requieren <strong>intervención inmediata</strong> y <strong>monitoreo intensivo 24/7</strong>.
                </p>
            </div>
        </div>
    </div>
</div>
"""))
```
